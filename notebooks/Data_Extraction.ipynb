```python
import re
import pandas as pd
from tqdm import tqdm #python library to show progress meeters

def load_json_image_mapping(path, max_rows=None):
  rows = []
  current_influencer = None

  with open(path, "r", encoding="utf-8", errors="replace") as f:
    for line_num, line in enumerate(tqdm(f)):
      line = line.strip()

      if not line:
        continue

      # skip the header line and//or separator lines
      if "influencer_name" in line or line.startswith("="):
        continue

      # stop early if just testing
      if max_rows is not None and len(rows) >= max_rows:
        break

      parts = line.split()

      # Case 1: Influencer name is present
      # Ex: 00_rocketgirl 123413241234231423423.info ['1234123412341234.jpg']
      if len(parts) >= 3 and parts[1].endswith(".info"):
        current_influencer = parts[0]
        json_file = parts[1]
        image_file = " ".join(parts[2:])

      # Case 2: influencer name is missing, so just reuse previous influenc
      # Ex: 11234123412341234.info ['123432413241324.jpg']
      elif len(parts) >= 2 and parts[0].endswith(".info"):
        json_file = parts[0]
        image_file = " ".join(parts[1:])

      else:
        continue

      rows.append({
          "influencer_name" : current_influencer,
          "json_file" : json_file,
          "image_file" : image_file
      })

  return pd.DataFrame(rows)
```

```python
import re
import pandas as pd

metadata_paths_df = pd.DataFrame({
    "metadata_path": [str(p) for p in sample_metadata_files],
    "extracted_file_name": [p.name for p in sample_metadata_files]
})

def extract_json_name(filename):
    match = re.search(r"(\d+\.info)$", filename)
    if match:
        return match.group(1)
    match = re.search(r"(\d+\.json)$", filename)
    if match:
        return match.group(1)
    return filename

metadata_paths_df["json_postmetadata_file_name"] = metadata_paths_df["extracted_file_name"].apply(extract_json_name)

metadata_paths_df.head()
```

```python
import json
import re
import pandas as pd
from tqdm import tqdm

def safe_get_count(data, key):
    """
    Handles nested Instagram count fields like:
    edge_media_preview_like: {"count": 182}
    """
    value = data.get(key, {})
    if isinstance(value, dict):
        return value.get("count", 0)
    return 0

def extract_caption(data):
    """
    Captions are usually stored in:
    edge_media_to_caption -> edges -> node -> text
    """
    caption_obj = data.get("edge_media_to_caption", {})
    edges = caption_obj.get("edges", [])

    if len(edges) > 0:
        node = edges[0].get("node", {})
        return node.get("text", "")

    return ""

def extract_location(data):
    location = data.get("location")
    if isinstance(location, dict):
        return location.get("name", None)
    return None

def extract_sponsor_count(data):
    sponsor_obj = data.get("edge_media_to_sponsor_user", {})
    if isinstance(sponsor_obj, dict):
        edges = sponsor_obj.get("edges", [])
        return len(edges)
    return 0

def extract_hashtags(text):
    if not isinstance(text, str):
        return []
    return re.findall(r"#(\w+)", text.lower())

def parse_metadata_file(path):
    try:
        with open(path, "r", encoding="utf-8", errors="replace") as f:
            data = json.load(f)
    except Exception:
        return None

    caption = extract_caption(data)

    return {
        "post_id": data.get("id"),
        "owner_username": data.get("owner", {}).get("username") if isinstance(data.get("owner"), dict) else None,
        "shortcode": data.get("shortcode"),
        "likes": safe_get_count(data, "edge_media_preview_like"),
        "comments": safe_get_count(data, "edge_media_to_parent_comment"),
        "preview_comments": safe_get_count(data, "edge_media_preview_comment"),
        "caption": caption,
        "caption_length": len(caption) if isinstance(caption, str) else 0,
        "hashtags": extract_hashtags(caption),
        "num_hashtags": len(extract_hashtags(caption)),
        "timestamp": data.get("taken_at_timestamp"),
        "is_ad": data.get("is_ad", False),
        "sponsor_count": extract_sponsor_count(data),
        "is_video": data.get("is_video", False),
        "post_type": data.get("__typename"),
        "location_name": extract_location(data),
        "width": data.get("dimensions", {}).get("width") if isinstance(data.get("dimensions"), dict) else None,
        "height": data.get("dimensions", {}).get("height") if isinstance(data.get("dimensions"), dict) else None,
    }
```

```python
posts_sample_df["datetime"] = pd.to_datetime(
    posts_sample_df["timestamp"],
    unit="s",
    errors="coerce"
)

posts_sample_df["hour"] = posts_sample_df["datetime"].dt.hour
posts_sample_df["day_of_week"] = posts_sample_df["datetime"].dt.day_name()
posts_sample_df["month"] = posts_sample_df["datetime"].dt.month

def hour_bucket(hour):
    if pd.isna(hour):
        return "unknown_time"
    elif 5 <= hour < 12:
        return "morning"
    elif 12 <= hour < 17:
        return "afternoon"
    elif 17 <= hour < 22:
        return "evening"
    else:
        return "night"

posts_sample_df["time_bucket"] = posts_sample_df["hour"].apply(hour_bucket)

posts_sample_df[["timestamp", "datetime", "hour", "time_bucket", "day_of_week"]].head()
```

```python
def caption_bucket(length):
    if pd.isna(length):
        return "unknown_caption"
    elif length < 80:
        return "short_caption"
    elif length < 200:
        return "medium_caption"
    else:
        return "long_caption"

def hashtag_bucket(n):
    if pd.isna(n):
        return "unknown_hashtags"
    elif n == 0:
        return "no_hashtags"
    elif n <= 3:
        return "few_hashtags"
    else:
        return "many_hashtags"

posts_sample_df["caption_bucket"] = posts_sample_df["caption_length"].apply(caption_bucket)
posts_sample_df["hashtag_bucket"] = posts_sample_df["num_hashtags"].apply(hashtag_bucket)
posts_sample_df["ad_bucket"] = posts_sample_df["is_ad"].apply(lambda x: "ad" if x else "not_ad")
posts_sample_df["media_bucket"] = posts_sample_df["is_video"].apply(lambda x: "video" if x else "image")

posts_sample_df["strategy"] = (
    posts_sample_df["time_bucket"].astype(str)
    + " + " + posts_sample_df["caption_bucket"].astype(str)
    + " + " + posts_sample_df["hashtag_bucket"].astype(str)
    + " + " + posts_sample_df["ad_bucket"].astype(str)
    + " + " + posts_sample_df["media_bucket"].astype(str)
)

posts_sample_df[[
    "influencer_name",
    "likes",
    "comments",
    "caption_bucket",
    "hashtag_bucket",
    "time_bucket",
    "ad_bucket",
    "media_bucket",
    "strategy"
]].head()
```

# build first baseline recommender

```python
def recommend_global_strategies(strategy_scores, k=5, min_posts=10):
    """
    Recommends the highest-performing content strategies overall.
    This is our first baseline recommender.
    """
    filtered = strategy_scores[strategy_scores["post_count"] >= min_posts]

    return filtered.sort_values(
        "avg_engagement",
        ascending=False
    ).head(k)

recommend_global_strategies(strategy_scores, k=5, min_posts=10)
```

```python
category_strategy_scores = (
    posts_base_sample_df
    .groupby(["category", "strategy"])
    .agg(
        post_count=("strategy", "count"),
        avg_engagement=("log_engagement_score", "mean"),
        avg_likes=("likes", "mean"),
        avg_comments=("comments", "mean")
    )
    .reset_index()
    .sort_values(["category", "avg_engagement"], ascending=[True, False])
)

category_strategy_scores.head(10)
```

```python
def recommend_by_category(category, category_strategy_scores, k=5, min_posts=3):
    """
    Recommends strategies that performed well for influencers in the same category.
    """
    filtered = category_strategy_scores[
        (category_strategy_scores["category"] == category)
        & (category_strategy_scores["post_count"] >= min_posts)
    ]

    return filtered.sort_values(
        "avg_engagement",
        ascending=False
    ).head(k)

recommend_by_category("fashion", category_strategy_scores, k=5, min_posts=3)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
import numpy as np

# Fill missing values with 0 for similarity calculation
interaction_matrix_filled = interaction_matrix.fillna(0)

# Compute user-user similarity
user_similarity = cosine_similarity(interaction_matrix_filled)

user_similarity_df = pd.DataFrame(
    user_similarity,
    index=interaction_matrix_filled.index,
    columns=interaction_matrix_filled.index
)

user_similarity_df.head()
```

```python
def recommend_user_based_cf(
    influencer_name,
    interaction_matrix,
    user_similarity_df,
    k=5,
    n_neighbors=10
):
    """
    Recommend strategies for one influencer using user-based collaborative filtering.
    """

    if influencer_name not in interaction_matrix.index:
        return pd.DataFrame({"error": [f"{influencer_name} not found in interaction matrix"]})

    # Similar influencers
    similarities = user_similarity_df[influencer_name].drop(index=influencer_name)
    top_neighbors = similarities.sort_values(ascending=False).head(n_neighbors)

    # Keep only positive similarities
    top_neighbors = top_neighbors[top_neighbors > 0]

    if len(top_neighbors) == 0:
        return pd.DataFrame({"error": ["No similar influencers found"]})

    # Neighbor strategy scores
    neighbor_scores = interaction_matrix.loc[top_neighbors.index]

    # Weighted average score for each strategy
    weighted_scores = neighbor_scores.T.dot(top_neighbors)
    similarity_sum = top_neighbors.sum()

    predicted_scores = weighted_scores / similarity_sum

    # Remove strategies the influencer already has
    already_used = interaction_matrix.loc[influencer_name].dropna().index
    predicted_scores = predicted_scores.drop(index=already_used, errors="ignore")

    recommendations = (
        predicted_scores
        .sort_values(ascending=False)
        .head(k)
        .reset_index()
    )

    recommendations.columns = ["recommended_strategy", "predicted_score"]

    return recommendations
```
